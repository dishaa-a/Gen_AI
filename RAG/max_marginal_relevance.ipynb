```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
# 1. Documents defined 
documents = [
    Document(
        page_content="Python is a high-level, interpreted programming language known for its readability and simple syntax. It supports multiple programming paradigms including procedural, object-oriented, and functional programming.",
        metadata={"source": "doc1", "topic": "programming"}
    ),
    Document(
        page_content="Machine learning is a subset of artificial intelligence that enables systems to learn patterns from data without being explicitly programmed. Common algorithms include decision trees, neural networks, and support vector machines.",
        metadata={"source": "doc2", "topic": "ai"}
    ),
    Document(
        page_content="Deep learning uses multi-layered neural networks to model complex patterns in data. It has driven major advances in image recognition, natural language processing, and speech recognition.",
        metadata={"source": "doc3", "topic": "ai"}
    ),
    Document(
        page_content="The Great Barrier Reef is the world's largest coral reef system, located off the coast of Queensland, Australia. It is composed of over 2,900 individual reefs and is visible from outer space.",
        metadata={"source": "doc4", "topic": "geography"}
    ),
    Document(
        page_content="Climate change refers to long-term shifts in temperatures and weather patterns, largely driven by human activities such as burning fossil fuels. It has significant effects on ecosystems, sea levels, and biodiversity.",
        metadata={"source": "doc5", "topic": "environment"}
    ),
    Document(
        page_content="Natural language processing (NLP) is a field of AI focused on enabling computers to understand, interpret, and generate human language. Applications include chatbots, translation, and sentiment analysis.",
        metadata={"source": "doc6", "topic": "ai"}
    ),
]
```

```python
# 2. Embeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)
```

```python
# 3. Vector store
vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    persist_directory="./chroma_db"
)
```

```python
# 4. MMR retriever — returns 5 documents
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 3,
        "fetch_k": 6,       # pool size (limited here since we only have 6 docs)
        "lambda_mult": 0.5
    }
)
```

```python
# 5. Query
query = "Tell me about artificial intelligence and machine learning"
results = retriever.invoke(query)

for i, doc in enumerate(results):
    print(f"--- Result {i+1} ---")
    print(doc.page_content)
    print(doc.metadata)
    print()
```
