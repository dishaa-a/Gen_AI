```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)
```

```python
documents = [
    Document(
        page_content="""
        Artificial Intelligence (AI) is a field of computer science focused on
        creating systems that can perform tasks that normally require human
        intelligence. These tasks include reasoning, learning, problem solving,
        perception, and understanding natural language.
        """,
        metadata={"source": "AI Introduction"}
    ),

    Document(
        page_content="""
        Natural Language Processing (NLP) is a branch of artificial intelligence
        that focuses on enabling computers to understand, interpret, and generate
        human language. NLP is used in chatbots, translation systems, search
        engines, sentiment analysis, and voice assistants.
        """,
        metadata={"source": "NLP Introduction"}
    ),

    Document(
        page_content="""
        Machine Learning is a subset of artificial intelligence in which
        computers learn patterns from data instead of being explicitly
        programmed for every task. Common approaches include supervised,
        unsupervised, and reinforcement learning.
        """,
        metadata={"source": "Machine Learning"}
    ),

    Document(
        page_content="""
        Deep Learning uses neural networks with multiple layers to learn complex
        patterns from large amounts of data. It has applications in computer
        vision, speech recognition, natural language processing, and generative AI.
        """,
        metadata={"source": "Deep Learning"}
    ),

    Document(
        page_content="""
        Computer vision allows computers to process and understand visual
        information such as images and videos. It is commonly used for object
        detection, facial recognition, medical image analysis, and autonomous
        vehicles.
        """,
        metadata={"source": "Computer Vision"}
    )
]

print(f"Number of documents: {len(documents)}")
```

```python
vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings
)
```

```python
base_retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)
```

```python
query = "How does AI relate to understanding language?"

results = base_retriever.invoke(query)

for i, doc in enumerate(results):
    print(f"--- Result {i + 1} ---")
    print(doc.page_content)
    print(doc.metadata)
    print()
```
