```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
```

```python
# 1. Initialize Embeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# 2. Define Documents (This step was missing in your error)
docs = [
    Document(page_content="Chroma is a vector database for AI applications."),
    Document(page_content="Hugging Face provides open-source embedding models.")
]
```

```python

# 3. Create Vector Store (Now 'docs' is defined)
vector_store = Chroma.from_documents(
    documents=docs,
    embedding=embeddings,
    persist_directory='chroma_db',
    collection_name="hf_collection"
)
```

```python
# 4. Test Query
results = vector_store.similarity_search("What is Chroma?", k=1)
print(results[0].page_content)
```

```python
#add documents
vector_store.add_documents(docs)
```

```python
# view documents
vector_store.get(
    include=['embeddings', 'documents', 'metadatas']
)
```

```python
# search document
vector_store.similarity_search(
    query = 'Which one is an embedding model',
    k = 1
)
```

```python
# update document
updated_docs = Document(
    page_content="Chroma is something between a vector store and vector database.",
    metadata = {"model": "Application"}
)

vector_store.update_document(document_id='39744772-ab20-4839-81a1-ff925cd02d8b', document = updated_docs)
```

```python
# view documents
vector_store.get(
    include=['embeddings', 'documents', 'metadatas']
)
```
